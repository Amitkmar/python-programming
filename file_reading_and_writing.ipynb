```python
#aise data set ko hmm JSON data set bhi bolte hai
#JSON= java script object notation
data={
    "name" : "Amit",
    "mail_id" : "amit@gmail.com",
    "phone" : 987654310,
    "subject" : "Python"

}
```

```python
#aise data ki file ko read write kaise kare
#dump function data ko write krne main help krega
import json
with open("data.json","w") as f:
    json.dump(data,f)
```

```python
#load function is to read the data
with open("data.json","r") as f:
    data1=json.load(f)
```

```python
#how to read data from csv file
#comma seperated value
import csv
data=[["name","emai","phone_number"],
      ["Amit","amit@gmail.com",76845845],
      ["Aman","aman@gmail.com",92135845],
      ["Nitin","nitin@gmail.com",80545845]
      ]
```

```python
#how to write this data in csv file
with open("data.csv","w") as f:
    writer=csv.writer(f)
    for i in data:
        writer.writerow(i)
```

```python
#read this data
with open("data.csv","r") as f:
    read_data=csv.reader(f)
    for i in read_data:
        print(i)
```

```python
#binary data ko read wrie kaise kare
#kisi bhi image,video,audio ko text data ki form main open kre
#tab jo data hme dikhta hai wo binary form main hota hai
with open("test4.bin","wb") as f:
    f.write(b"\x01\x02\x03")
    #b matlab binary data
```

```python
#how to read
with open("test4.bin","rb") as f:
    print(f.read())
```
